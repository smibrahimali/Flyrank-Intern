## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (e.g., "refresh if age > 12 months") is too rigid and wastes resources on evergreen content that is still performing perfectly. ML captures the messy, non-linear interactions between publish age, recent CTR degradation, and seasonal search momentum—identifying pages that are actively losing ground regardless of their arbitrary age.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Demonstrating the complexity that breaks if-statements
feature_space = ['publish_age', 'ctr_degradation_rate', 'seasonal_momentum', 'competitor_density']

print("A fixed rule fails because it cannot simultaneously weigh the non-linear interactions between:")
for feature in feature_space:
    print(f"- {feature}")
```
