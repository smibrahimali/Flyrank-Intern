## 1. Question

*The research question and the decision it supports.*

**Predicting SEO content decay before traffic loss occurs.**
The core decision this supports is editorial triage. Instead of content teams guessing which pages to update or waiting for dashboards to show a traffic drop, this pipeline identifies high-value URLs that require immediate intervention versus those that are safe to monitor.

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

**Source:** FlyRank anonymized search telemetry dataset (representing the full 79M row release, utilizing the 30,000-row starter slice for local execution).
**Exclusions & Safety:** All private identifiers including client names, domains, URLs, page titles, and specific search queries were strictly excluded from the ingestion layer to guarantee public safety and prevent data leakage. The dataset relies entirely on aggregate behavioral metrics and temporal metadata.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

*   **Target Label:** `is_declining_label` (binary flag indicating an impending, significant drop in organic clicks).
*   **Features:** Rolling velocity metrics, specifically `days_with_impressions`, `log_impressions_90d`, `avg_position`, and `content_age_days`.
*   **Baseline:** A naive heuristic rule-set based on recency and basic thresholding.
*   **Validation Design (Leak-Free):** Implemented a strict `client_holdout` split strategy. Random k-fold cross-validation was explicitly avoided to prevent temporal leakage and ensure that historical features for a specific entity did not bleed across the train/validation boundary.

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

The models were evaluated on the exact same held-out validation split. The Random Forest classifier was selected as the optimal model based on its superior Precision@50 metric.

| Model | ROC-AUC | Avg Precision | Precision@50 | Recall | F1 Score |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **Random Forest (Selected)** | **0.750** | **0.618** | **0.740** | **0.744** | **0.640** |
| **Baseline Rules** | 0.627 | 0.468 | 0.240 | -- | -- |

*Observation:* The machine learning pipeline achieved a ~3x lift in precision for the top 50 prioritized URLs compared to the baseline hand-written rules.

## 5. Limitations

*What this work cannot claim.*

*   **Statistical, not Causal:** This model calculates the statistical probability of decay based strictly on historical user behavior telemetry.
*   **Unobservable Shocks:** It cannot predict external variables such as sudden, unannounced search engine core algorithm updates or sudden macroeconomic shifts in user search intent.
*   **Human-in-the-Loop:** The outputs are directional and decision-support tools; they require manual editorial verification before executing content changes.

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

The pipeline translates raw probabilities into a structured action playbook for reviewers:
1.  `refresh`: High decay risk; requires immediate editorial content updates.
2.  `refresh_and_review_ctr`: URL has high visibility but failing engagement; requires A/B testing of meta descriptions and titles.
3.  `monitor`: Content is performing stably; no intervention required.
