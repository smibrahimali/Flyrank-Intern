## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

Ranked Actions & Reason Codes:
The model output is mapped to clear archetypes to provide editorial teams with actionable next steps rather than just a raw probability score.
R1 (Urgent Refresh): High decay probability on high-volume, aged content (>365 days). Action: Full editorial review and content update.
R2 (Metadata Tweak): Moderate decay probability on mid-lifecycle content. Action: A/B test a new title or meta description to combat CTR fatigue.
R3 (Evergreen/Monitor): Low decay probability. Action: Do nothing; leave as is.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np
import os

# Simulate the scored output from the Week 6 validation model
np.random.seed(42)
n_pages = 100
df_scored = pd.DataFrame({
    'url': [f'/blog/article-{i}' for i in range(n_pages)],
    'publish_age_days': np.random.randint(50, 800, n_pages),
    'impressions_30d': np.random.randint(500, 50000, n_pages),
    'decay_probability': np.random.uniform(0.0, 1.0, n_pages)
})

# Map probabilities and features to Reason Codes
def assign_action(row):
    if row['decay_probability'] > 0.75 and row['publish_age_days'] > 365 and row['impressions_30d'] > 10000:
        return 'R1: Urgent Refresh'
    elif row['decay_probability'] > 0.50:
        return 'R2: Metadata Tweak'
    else:
        return 'R3: Monitor (Evergreen)'

df_scored['action_code'] = df_scored.apply(assign_action, axis=1)

# Rank the queue by highest business value (High Impressions + High Decay Risk)
df_queue = df_scored[df_scored['action_code'] != 'R3: Monitor (Evergreen)'].copy()
df_queue['priority_score'] = df_queue['decay_probability'] * np.log1p(df_queue['impressions_30d'])
df_queue = df_queue.sort_values(by='priority_score', ascending=False).reset_index(drop=True)

print("Top 5 Actionable Content Pieces:")
display(df_queue.head())
```
